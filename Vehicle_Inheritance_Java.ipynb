# Vehicle Manufacturing System – Java Inheritance

This notebook contains the Java program for `Vehicle`, `TwoWheeler`, `FourWheeler`, and `Electric_car`.

**Note:** This is Java source code, not Python. To execute it in Jupyter Notebook, use a Java kernel such as IJava. It can also be copied into Programiz or another Java compiler.

```java
class Vehicle {
    String brand, color, model;

    void getVehicle(String b, String c, String m) {
        brand = b;
        color = c;
        model = m;
    }

    void displayVehicle() {
        System.out.println("Brand: " + brand);
        System.out.println("Color: " + color);
        System.out.println("Model: " + model);
    }
}

class TwoWheeler extends Vehicle {
    double mileage;
    int cc;

    void display() {
        System.out.println("\n--- TWO WHEELER ---");
        displayVehicle();
        System.out.println("Mileage: " + mileage);
        System.out.println("CC: " + cc);
    }
}

class FourWheeler extends Vehicle {
    int cc;
    double weight;
    int seats;

    void display() {
        System.out.println("\n--- FOUR WHEELER ---");
        displayVehicle();
        System.out.println("CC: " + cc);
        System.out.println("Weight: " + weight);
        System.out.println("Seats: " + seats);
    }
}

class Electric_car extends FourWheeler {
    int battery;
    double power;
    String chargingType;

    void displayElectric() {
        System.out.println("\n--- ELECTRIC CAR ---");
        displayVehicle();
        System.out.println("CC: " + cc);
        System.out.println("Weight: " + weight);
        System.out.println("Seats: " + seats);
        System.out.println("Battery: " + battery + " kWh");
        System.out.println("Power: " + power + " kW");
        System.out.println("Charging Type: " + chargingType);
    }
}

public class Main {
    public static void main(String[] args) {

        TwoWheeler bike = new TwoWheeler();
        bike.getVehicle("Honda", "Black", "Shine");
        bike.mileage = 55.5;
        bike.cc = 125;

        FourWheeler car = new FourWheeler();
        car.getVehicle("Toyota", "White", "Innova");
        car.cc = 2500;
        car.weight = 1800;
        car.seats = 7;

        Electric_car ecar = new Electric_car();
        ecar.getVehicle("Tata", "Blue", "Nexon EV");
        ecar.cc = 0;
        ecar.weight = 1500;
        ecar.seats = 5;
        ecar.battery = 40;
        ecar.power = 100;
        ecar.chargingType = "Fast Charging";

        bike.display();
        car.display();
        ecar.displayElectric();
    }
}
```
